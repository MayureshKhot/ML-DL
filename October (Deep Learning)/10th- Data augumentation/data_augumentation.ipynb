```python
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.preprocessing.image import load_img, img_to_array
```

```python
# Create an image data generator with augumentations
datagen = ImageDataGenerator(
    rotation_range=40,
    width_shift_range=0.4,
    height_shift_range=0.4,
    shear_range=0.4,
    zoom_range=0.4,
    horizontal_flip=True,
    rescale=None,
    fill_mode='reflect' # other params: nearest, constant, wrap
)

img = load_img(r"D:\NareshIT_PrakashSenapati\Prakash Sir-Notes\October (Deep Learning)\10th- Data augumentation\img\1.jpg")
```

```python
img
```

```python
# convert to numpy array

x=img_to_array(img)
x=x.reshape((1,)+x.shape) #Reshape to (1, height, width, channels)

i = 0
for batch in datagen.flow(x, batch_size=1,
                          save_to_dir=r"D:\NareshIT_PrakashSenapati\Prakash Sir-Notes\October (Deep Learning)\10th- Data augumentation\img",
                          save_prefix='ironman', save_format='jpeg'):
    i += 1
    if i > 30:
        break # Prevent infinite loop
```
