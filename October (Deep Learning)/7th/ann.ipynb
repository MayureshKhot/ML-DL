```python
dataset = pd.read_csv(r"D:\NareshIT_PrakashSenapati\Prakash Sir-Notes\October (Deep Learning)\7th\Churn_Modelling.csv")
```

```python
dataset.head()
```

```python
X = dataset.iloc[:, 3:-1]  # from column index 3 up to but not including the last one
y = dataset.iloc[:, -1] 
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
X['Gender'] = le.fit_transform(X['Gender'])
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
ct = ColumnTransformer(transformers=[('encoder', OneHotEncoder(), [1])], remainder='passthrough')
X = np.array(ct.fit_transform(X))
```

```python
from sklearn.preprocessing import StandardScaler
st = StandardScaler()
X = st.fit_transform(X)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)
```

```python
print(X_train.shape)
print(X_test.shape)
print(y_train.shape)
print(y_test.shape)
```

# ANN Model Building Architechture

```python
ann = tf.keras.models.Sequential() # establish layer one after another
```

```python
ann.add(tf.keras.layers.Dense(units=6)) # input layer
```

```python
ann.add(tf.keras.layers.Dense(units=6, activation='relu')) #hidden layer 1
```

```python
ann.add(tf.keras.layers.Dense(units=5, activation='relu')) # hidden layer 2
```

```python
ann.add(tf.keras.layers.Dense(units=4, activation='relu')) # hidden layer 3
```

```python
ann.add(tf.keras.layers.Dense(units=1, activation='sigmoid')) # output layer
```

```python
ann.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
ann.fit(X_train, y_train, batch_size=32, epochs=50)
```
