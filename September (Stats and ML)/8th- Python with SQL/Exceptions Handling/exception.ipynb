# Execption handling in Python

- Compile time error is syntax error
- It's made by user

## Logical Error
- 2+4=5
- These are logical error, cos logic fails here

## Runtime error
- Error occured at runtime is called as runtime error
- 5/0

## Try, Except, Finally
- statement in try block is called as Critical Statement.
- If try block throws error, then interpreter goes to except
- If no error, then except block is skipped.
- Finally block executes no matter what other code does.
